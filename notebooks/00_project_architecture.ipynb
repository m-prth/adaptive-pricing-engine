# The Core Concept: "Dual-Brain" Architecture

Traditional lending models only answer one question: _"Will they pay us back?_ (Risk).This project adds a second brain: _"Will they accept the offer?"_ (Demand).

By combining these two probabilities, we move from **Risk-Based Pricing** (static tables) to **Profit-Optimized Pricing** (dynamic curves).

**The Logical Flow**
1. Inputs: Customer Application + Historical Data.

2. The "Causal" Fix: We generate Synthetic Data to simulate a Randomized Control Trial (RCT), allowing the model to learn price elasticity without selection bias.

3. Brain 1 (Risk): XGBoost Classifier predicts Probability of Default ($PD$).
    
4. Brain 2 (Demand): Logistic Regression predicts Probability of Acceptance ($P(Accept)$).

5. Optimization: We find the interest rate $r$ that maximizes:$$E[\text{Profit}] = P(\text{Accept}|r) \times \left[ (1-PD) \times \text{Interest Income}(r) - PD \times \text{Loss} \right]$$

6. Governance: A deterministic policy layer overrides the AI to ensure regulatory compliance (e.g., Usury Caps, Fair Lending).

# Global Configuration & Guardrails

This section defines the constants used throughout the project. Changing a value here updates the behavior of the entire pipeline (e.g., tightening the PD threshold).

| Constant Name | Value | Description |
| --- | --- | --- |
| RISK_MODEL_PATH | `../models/risk_model_xgb.pkl` | File path for the saved XGBoost risk model artifact. |
| ELASTICITY_MODEL_PATH | `../models/elasticity_model_logit.pkl` | File path for the saved logistic regression elasticity model artifact. |
| COST_OF_FUNDS | 4% | The cost for the bank to borrow money (4%), used in profit calculations. |
| LGD | 60% | Loss Given Default; assumes a 60% principal loss if a borrower defaults. |
| GLOBAL_MAX_RATE | 35% | The absolute maximum interest rate allowed (Usury Limit). |
| GLOBAL_MIN_RATE | 5% | The minimum interest rate required to cover operational costs. |
| MAX_PD_THRESHOLD | 20% | Risk appetite threshold; applicants with a Probability of Default > 20% are auto-rejected. |
| PRIME_MAX_RATE | 20% | Interest rate cap for "Prime" (high FICO) borrowers to protect brand reputation. |
| MIN_PROFIT_MARGIN | $50 | Minimum Expected Profit ($50) required to approve a loan. |
| ELASTICITY_ALPHA | 30 | Coefficient controlling price sensitivity in synthetic data generation; higher values mean a steeper demand curve. |
